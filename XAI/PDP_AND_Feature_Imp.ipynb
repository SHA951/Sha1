```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.inspection import PartialDependenceDisplay
import matplotlib.pyplot as plt
```

```python
data = fetch_california_housing()
```

```python
x = pd.DataFrame(data.data,columns=data.feature_names)
y = data.target
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
model = RandomForestRegressor(n_estimators=100)
model.fit(x_train,y_train)
```

```python
PartialDependenceDisplay.from_estimator(model,x_train,["MedInc"],feature_names=data.feature_names)
plt.show()
```

```python
PartialDependenceDisplay.from_estimator(model,x_train,[("MedInc","AveRooms")],feature_names=data.feature_names)
plt.show()
```

```python
imp_feat = model.feature_importances_
fea  = x_train.columns
import seaborn as sns
import matplotlib.pyplot as plt
sns.barplot(x=imp_feat,y=fea)
plt.show()
```
