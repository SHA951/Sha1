```python
df = df.rename(columns = str.upper).head()
df
```

```python
df.describe()
```

```python
cols_to_convert = df.columns[1:]
df[cols_to_convert] = df[cols_to_convert].apply(pd.to_numeric,errors = "coerce")
df.fillna(method = "ffill", inplace = True)

plt.figure(figsize = (10,6))
sns.boxplot(df)
plt.xticks(rotation = 90)
plt.title("Outliers detection using Box Plot")
plt.show()
```

```python
X = df["GLUCOSE"]
Y = df["CLASS"]

plt.scatter(X,Y)
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(df[['PREGNANT', 'GLUCOSE', 'DIASTOLIC_BP', 'SKIN_FOLD', 'SERUM_INSULIN',
       'BMI', 'DIABETES_PEDIGREE', 'AGE', 'CLASS']])
plt.suptitle('Pair Plot of Sector Growth Rates', y=1.02)
plt.show()
```
