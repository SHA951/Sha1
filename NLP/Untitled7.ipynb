```python
emission_probs = {
    'x': {'A': 0.6, 'B': 0.2, 'C': 0.2},
    'y': {'A': 0.1, 'B': 0.1, 'C': 0.8},
    'z': {'A': 0.5, 'B': 0.2, 'C': 0.3}
}

transition_probs = {
    'x': {'x': 0.8, 'y': 0.2, 'z': 0.0},
    'y': {'x': 0.0, 'y': 0.2, 'z': 0.8},
    'z': {'x': 0.0, 'y': 0.0, 'z': 1}
}

alpha_x=1
alpha_y=0
alpha_z=0

sequence=['B','A','B']


alpha_values={'X':[],'Y':[],'Z':[]}

for obs in sequence:
    old_alpha_x=alpha_x
    old_alpha_y=alpha_y
    old_alpha_z=alpha_z
    alpha_x = (old_alpha_x * transition_probs['x']['x'] * emission_probs['x'][obs] +
               old_alpha_y * transition_probs['y']['x'] * emission_probs['x'][obs] +
               old_alpha_z * transition_probs['z']['x'] * emission_probs['x'][obs])

    alpha_y = (old_alpha_x * transition_probs['x']['y'] * emission_probs['y'][obs] +
               old_alpha_y * transition_probs['y']['y'] * emission_probs['y'][obs] +
               old_alpha_z * transition_probs['z']['y'] * emission_probs['y'][obs])

    alpha_z = (old_alpha_x * transition_probs['x']['z'] * emission_probs['z'][obs] +
               old_alpha_y * transition_probs['y']['z'] * emission_probs['z'][obs] +
               old_alpha_z * transition_probs['z']['z'] * emission_probs['z'][obs])

    alpha_values['X'].append(alpha_x)
    alpha_values['Y'].append(alpha_y)
    alpha_values['Z'].append(alpha_z)

print("Alpha values for each state at each time step:")
print("X:",alpha_values['X'])
print("Y:",alpha_values['Y'])
print("Z:",alpha_values['Z'])

probability_sequence= alpha_x + alpha_y + alpha_z
print("Probability of the sequence {BAB}:", probability_sequence)

      
```

```python
emission_probs = {
    'x': {'A': 0.6, 'B': 0.2, 'C': 0.2},
    'y': {'A': 0.1, 'B': 0.1, 'C': 0.8},
    'z': {'A': 0.5, 'B': 0.2, 'C': 0.3}
}

transition_probs = {
    'x': {'x': 0.8, 'y': 0.2, 'z': 0.0},
    'y': {'x': 0.0, 'y': 0.2, 'z': 0.8},
    'z': {'x': 0.0, 'y': 0.0, 'z': 1}
}

sequence=['B','A','C']
n=len(sequence)

beta_x=[0]*n
beta_y=[0]*n
beta_z=[0]*n


beta_x[-1] = 1
beta_y[-1] = 1
beta_z[-1] = 1

# Backward procedure
for t in range(n - 2, -1, -1):
    # Update beta values for each state at time step t
    beta_x[t] = (
        transition_probs['x']['x'] * emission_probs['x'][sequence[t + 1]] * beta_x[t + 1] +
        transition_probs['x']['y'] * emission_probs['y'][sequence[t + 1]] * beta_y[t + 1]
    )
    beta_y[t] = (
        transition_probs['y']['y'] * emission_probs['y'][sequence[t + 1]] * beta_y[t + 1] +
        transition_probs['y']['z'] * emission_probs['z'][sequence[t + 1]] * beta_z[t + 1]
    )
    beta_z[t] = (
        transition_probs['z']['z'] * emission_probs['z'][sequence[t + 1]] * beta_z[t + 1]
    )

# Initial probability of starting at state 'x' and observing the sequence
initial_prob = 1  # Given that the machine always starts in state x
final_probability = (
    initial_prob * emission_probs['x'][sequence[0]] * beta_x[0]
)

# Print beta values
print("Beta values for state x:", beta_x)
print("Beta values for state y:", beta_y)
print("Beta values for state z:", beta_z)
print("Probability of the sequence {BAB} using backward procedure:", final_probability)
```
