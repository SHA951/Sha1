```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
def preprocess_text(text,use_stemmer=True):
    words = word_tokenize(text.lower())
    stopwords = set(stopwords.words('english'))
    porter_stemmer = PorterStemmer()
    wordnet_lemmatizer = WordNetLemmatizer()
    filtered_words = [word for word in words if word not in stopwords and word.isalpha()]
    
    if use_stemmer:
        stemmed_words = [porter_stemmer.stem(word) for word in words]
        return " ".join(stemmed_words)
    else:
        lemmatized_words = [wordnet_lemmatizer.lemmatize(word) for word in words]
        return " ".join(lemmatized_words)
    
df["Processed_stemmed"]=df['Text'].apply(preprocess_text)
df["Processed_lemmatized"]=df['Text'].apply(lambda x: preprocess_text(x,use_stemmer=False))
```
