```python
with open(r'random_collocation_text.txt','r') as f:
    content=f.read()
    l=content.split("\n")
    for i in l:
        if i=="":
            l.remove("")

print(l)
```

```python
df=pd.DataFrame({"Text":l})
df.head()
```

```python
def preprocess(text,stemming=True):
    tokens=word_tokenize(text)
    stop_words=set(stopwords.words('english'))
    tokens= [token for token in tokens if token.lower() not in stop_words]

    if stemming:
        stemmer=PorterStemmer()
        tokens=[stemmer.stem(token) for token in tokens]
    else:
        Lemmatizer=WordNetLemmatizer()
        tokens=[Lemmatizer.lemmatize(token) for token in tokens]

    print(tokens)
    return tokens
```

```python
df['Preprocessed']=df['Text'].apply(preprocess)
```

```python
import math
def t_test_collocation(unigram_counts,bigram_counts,word1,word2,total_count):
    count_word1=unigram_counts.get(word1,0.00000000001)
    count_word2=unigram_counts.get(word2,0.000000000001)
    bigram_count=bigram_counts.get((word1,word2),0.00000000000001)

    expected_freq=(count_word1+count_word2)/total_count

    variance_word1 = sum((freq - count_word1) ** 2 for freq in unigram_counts.values()) / (len(unigram_counts) - 1)
    variance_word2=sum((freq-count_word2)**2 for freq in unigram_counts.values())/(len(unigram_counts)-1)

    standard_error = math.sqrt(variance_word1 / count_word1 + variance_word2 / count_word2)
    standard_error = math.sqrt(variance_word1 / count_word1 + variance_word2 / count_word2)
    
    # t-statistic
    t_stat = (bigram_count - expected_freq) / standard_error   #t=(x-u)/(se)
    
    return t_stat


t_test_collocation(unigram_dict,bigram_dict,"pradeepa","niggi",len(bigram_dict))
    
```
