```python
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
```

```python
df=pd.read_csv(r"E:\shampoo_sales.csv")
df.head(10)
```

```python
df['Month']=pd.to_datetime(df['Month'], format='%d-%m')
df
```

```python
df.set_index(df['Month'],inplace=True)
df
```

```python
df['Sales'].plot(kind='hist',bins=5,title='Shampoo sales')
ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)
plt.show()
```

```python
df['Sales'].plot(kind='hist',bins=5,title='Shampoo sales')
ax=plt.gca()
ax.spines['top'].set_visible(True)
ax.spines['right'].set_visible(True)
ax.spines['left'].set_visible(True)
ax.spines['bottom'].set_visible(True)
plt.show()
```

```python
df['Sales'].plot(kind='line',figsize=(8,4),title='shampoo sales')
ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
```

```python
plt.figure(figsize=(15,10))
plt.plot(df["Sales"])
plt.xlabel("Date")
plt.ylabel("Shampoo sales")
plt.title("Shampoo sales time series analysis")
plt.show()
```

```python
df['month']=df.index.month
df['day']=df.index.day
df
```

```python
df['lag_1']=df['Sales'].shift(1)
df
```

```python
df['lag_2']=df['Sales'].shift(2)
df
```

```python
df["expanding_mean"] = df["Sales"].expanding().mean()
df["expanding_std"] = df["Sales"].expanding().std()
df["expanding_min"] = df["Sales"].expanding().min()
df["expanding_max"] = df["Sales"].expanding().max()
df
```
