```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
values=np.random.randn(100)
```

```python
values
```

```python
dates=pd.date_range(start="2025-01-01", end="2025-04-10", freq="D")
```

```python
dates
```

```python
df=pd.DataFrame(values, index=dates, columns=['Value'])
```

```python
df
```

```python
df.plot(title=" Synthetic Data Plot", figsize=(14,8), legend=None, color='mediumaquamarine')
plt.xlabel('Date')
plt.ylabel('Synthetic Data')
plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

fig,axes=plt.subplots(2,1,figsize=(10,8))
plot_acf(df["Value"], lags=30, ax=axes[0], title="Autocorrelation Function (ACF)", color='mediumturquoise')
plot_pacf(df["Value"], lags=30, ax=axes[1], title="Partial Autocorrelation Function (PACF)", color='paleturquoise')
plt.tight_layout()
plt.show()
```
