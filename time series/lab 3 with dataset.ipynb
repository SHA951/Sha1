```python
df["Date"]=pd.to_datetime(df["Date"])
df
```

```python
df.set_index('Date',inplace=True)
```

```python
df.head(5)
```

```python
df=df.loc['1981-01-01':'1982-01-31']
df
```

```python
plt.figure(figsize=(11,3))
plt.plot(df.index,df['Temp'])
plt.title('Daily Temperature in jan 1981')
plt.xlabel('Date')
plt.ylabel('Temp in celcius')
plt.show()
```

```python
plt.figure(figsize=(11,4))
plt.scatter(df.index,df['Temp'])
plt.title('Daily Temperature in jan 1981')
plt.xlabel('Date')
plt.ylabel('Temp in celcius')
plt.show()
```

```python
plt.figure(figsize=(6,3))
pd.plotting.autocorrelation_plot(df['Temp'])
plt.title('Autocorrelation plot')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['Temp'],bins=5)
plt.title('Daily Temperature in jan 1981')
plt.xlabel('Date')
plt.ylabel('Temp in celcius')
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.kdeplot(df['Temp'],fill=True)
plt.title('Daily Temperature in jan 1981 density plot')
plt.xlabel('Temperature')
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.boxplot(x=df['Temp'])
plt.title('Daily Temperature in jan 1981 box plot')
plt.xlabel('Temperature')
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.heatmap(df,cmap='coolwarm')
plt.title('Heat map for Daily Temperature in jan 1981')
plt.xlabel('Temperature')
plt.show()
```
